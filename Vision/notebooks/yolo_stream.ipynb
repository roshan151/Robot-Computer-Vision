```python
import ffmpeg
def convert_mp4_to_mov(input_file, output_file):
    """Converts an MP4 file to MOV format."""

    try:
        ffmpeg.input(input_file).output(output_file, format='mov').run()
        print(f"Conversion successful: {input_file} -> {output_file}")
    except Exception as e:
        print(f"Error converting file: {e.stderr.decode()}")

input_file = "/Users/roshantiwari/src/Robot-Computer-Vision/vid.mp4"
output_file = "new_vid.mov"
convert_mp4_to_mov(input_file, output_file)
```
